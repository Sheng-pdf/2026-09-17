為自己學Python

安裝 Python 的困難度不高，基本上就是到 Python 的官方網站下載安裝程式並照著指示做就行了，目前主流的作業系統，例如 Windows、macOS 以及 Linux 都有相對應的版本。是說，如果各位使用的是 macOS 跟 Linux 作業系統，可能會發現原本電腦裡就已經直接內建 Python 了，這樣還需要另外安裝嗎？作業系統內建安裝的 Python 通常會比現行版本稍微舊一點，如果單純就以體驗或寫一些簡單的程式的話是夠用的。作業系統之所以會內建安裝 Python 是有原因的，因為有些系統程式本身就是用 Python 開發的。Python 的版本從早期的 2.x 版到現在的 3.x 版，除了正宗的 CPython 之外，還有其他分支的實作品版本（像是 Jython、IronPython、PyPy 等），如果不想把自己的電腦環境弄的亂七八糟，或是因為專案需要常常安裝不同版本的 Python，通常會建議另外使用版本管理工具來建構隔離的開發環境，這裡將使用 pyenv 做為範例。

可以把 pyenv 想像成一個工具箱，在這個工具箱裡會可以放很多種工具，但這個工具箱裡如果沒有任何工具它就只是個空的盒子而已。pyenv 不只是個工具箱，還是有分隔板的工具箱，不同版本的 Python 會被安裝在不同的隔板裡，安裝套件的時候也會安裝在各自的隔板裡不會互相干擾，透過 pyenv 就能做到簡單的切換 Python 的環境及版本。

提醒一下，如果你的作業系統（例如 macOS）原本就有預先安裝 Python 的話，即使透過 pyenv 安裝了其他版本的 Python，原本系統內建的 Python 也建議不要移除，它存在作業系統裡一定有原因的，你不會知道系統裡有哪些程式正在使用它。

pyenv 的主要功能包括：

- 安裝多個 Python 版本
- 切換全域（global）Python 版本
- 為不同專案指定 Python 版本
- 管理 CPython、PyPy 等不同 Python 發行版本

1. pyenv 的主要用途是什麼？

自由切換：隨時在不同專案之間切換 Python 版本（例如專案 A 用 3.9，專案 B 用 3.12）。

隔離系統：避免動到作業系統內建的 Python，防止系統壞掉。

一鍵安裝：直接用指令下載並安裝任何指定的 Python 版本。
2. 為什麼有些開發者會在同一台電腦安裝多個 Python 版本？

維護舊專案與開發新專案：舊專案可能需要舊版 Python（例如 3.8）才能運作，而新專案想使用新版（例如 3.12）的新功能。

測試套件的相容性：開發套件或套件庫時，需要確認程式碼在不同 Python 版本上都能正常執行。

避免破壞系統環境：作業系統本身可能依賴特定版本的 Python，單獨安裝其他版本可以避免把系統套件用壞。
3. 請說明下列指令的用途：

   (a) `pyenv versions`

 列出目前電腦中所有已安裝的 Python 版本，並會在目前正在使用的版本旁邊標示星號（*）。
   (b) `pyenv install 3.12.7`

從官方下載並安裝指定版本（此處為 3.12.7）的 Python 到你的電腦中。
   (c) `pyenv shell 3.12.7`

僅在目前的 Terminal 視窗中將 Python 切換為 3.12.7。關閉視窗或打開新分頁後就會失效（作用域最短、優先權最高）。
   (d) `pyenv global 3.12.7`

設定全域預設的 Python 版本。當你沒有針對特定目錄設定版本時，整台電腦預設都會使用 3.12.7。
   (e) `pyenv local 3.12.7`

將當前資料夾綁定使用 3.12.7。它會在該目錄下生成一個隱藏檔 .python-version，只要進入這個資料夾，系統就會自動切換到 3.12.7（適合個別專案使用）。

4. `shell`、`local` 與 `global` 三種設定方式有何差異？

shell：只影響現在打開的這個視窗（關掉就失效，優先權最高）。

local：只影響特定專案資料夾（進資料夾自動切換，優先權第二）。

global：影響全域預設（沒指定時全電腦都用它，優先權最低）。

5. pyenv 與 venv 有何不同？

pyenv 管「Python 本身」：負責在電腦裡下載、安裝、切換不同版本的 Python（例如 3.9、3.11、3.12）。

venv 管「套件（Packages）」：負責在同一個 Python 版本下，為個別專案建立獨立的套件隔離環境（例如專案 A 用 Django 4.0，專案 B 用 Django 5.0）

6. Git、pyenv 與 venv 分別負責管理什麼？

pyenv 管 「Python 要用哪一版」

venv 管 「套件要裝哪些」

Git 管 「程式碼怎麼改的」

7. 本課程使用 GitHub Codespaces，
   為什麼通常不需要使用 pyenv？

環境已經完全隔離：Codespaces 本身就是一個獨立執行的 Linux 容器（Docker Container），不是你的個人電腦，不需要擔心影響到其他專案或系統環境。

版本早已在設定檔定案：專案的 Python 版本通常已經在 .devcontainer/devcontainer.json 或 Dockerfile 中指定好了，啟動時即為正確版本。

需要換版本直接改設定即可：如果真的需要更換 Python 版本，只需修改 devcontainer 設定並重新建置（Rebuild）容器，幾秒鐘就能切換，不需要透過 pyenv 在容器內手動下載與切換。

8. 如果未來你在自己的電腦上同時維護：

   - 專案 A：Python 3.10
   - 專案 B：Python 3.13

   你會如何利用 pyenv 解決版本衝突？

先下載：用 pyenv install 3.10.14 與 3.13.0 把兩個版本裝好。

再綁定：分別進到兩個專案資料夾，輸入 pyenv local 3.10.14（專案 A）和 pyenv local 3.13.0（專案 B）。

建環境：在各自資料夾內執行 python -m venv .venv 建立獨立虛擬環境。

#### 套件和虛擬環境

在大多數其他的程式語言之中， 一個套件( package ) 相當於一個程式庫(library)，都是由一組程式碼組成。Python 因為「一切需要的都已包含在內」而聞名， 大部分功能只需透過簡單的import陳述式就能取用。但如果您需要處理一些超過基本功能的事情， 例如建置華麗的使用者介面， 通常就需要安裝另一個套件來配合。

還好大多數第三方程式庫的安裝都相當簡單。程式庫的作者把它們打包成套件，「pip 套件管理器」可以用來安裝這些套件。

使用多個第三方套件需要一些技巧。有些套件需要先安裝其他套件來配合， 有些套件還可能與其他套件衝突，我們還可以根據需要安裝特定版本的套件。之前是否有提到電腦上的某些應用程式和作業系統元件依賴於特定的Python套件呢？這就是為什麼虛擬環境存在的原因。

虛擬環境就像沙盒， 您可以在其中只安裝特定於某個專案所需的Python套件，避免這些套件與其他專案（或您的系統）產生衝突的風險。您可以為每個專案建立不同的小沙盒， 只在其中安裝需要的套件， 讓一切都能保持整齊有序。實際上我們並沒有改變系統上已安裝的Python 套件， 因此可以避免破壞與您的專案無關但又很重要的事物。

#### 建立虛擬環境

每個虛擬環境都放在一個專屬的目錄中。通常我們會把這個資料夾命名為env或venv。對於每個專案， 我通常喜歡在專案資料夾內建立一個專屬的虛擬環境。Python提供了一個名為venv 的工具來完成這個工作。如果您使用Git或其他版本控制系統( version control system, VCS ) 來追蹤程
式碼的變更， 還有一個額外的設定步驟在後面會提到。

若想要在目前的工作目錄中建立名為venv 的虛擬環境， 請在命令列中執行以下指令：
```bash
python -m venv .venv
```
第一個venv是一個用來建立虛擬環境的指令，而第二個venv是指定虛擬環境的路徑。在上述的範例中，venv 只是個相對路徑，會在目前的工作目錄
中建立一個「.venv/ 」目錄。不過，也可以使用絕對路徑 並且可以任意取名。現在如果査看專案的工作目錄， 就會看到「venv/ 」資料夾已經被建立了。

若使用 Git 或 GitHub 進行版本控制，通常不會將 .venv 目錄加入版本控制，而是將其加入 .gitignore：
```bash
.venv/
```
這是因為虛擬環境中包含大量可重新建立的檔案，不需要上傳至 GitHub。相反地，通常會將套件資訊記錄在 requirements.txt 中。
```bash
pip freeze > requirements.txt
```

其他開發者便可利用：
```bash
pip install -r requirements.txt
```
重新建立相同的套件環境。

#### 啟用虐擬環境（CLI）

建立虛擬環境後，必須先啟用（Activate）才能使用其中的 Python 解譯器與已安裝的套件。

- 在 Linux、macOS 或 GitHub Codespaces 中，可使用下列指令啟用虛擬環境：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows 的命令提示字元（Command Prompt）中，請使用：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows PowerShell 中，請使用：
```bash
.venv\Scripts\Activate.ps1
```
若 PowerShell 出現執行政策（Execution Policy）相關錯誤，可在 PowerShell 中執行：
```
Set-ExecutionPolicy RemoteSigned -Scope CurrentUser
```

啟用成功後，命令提示字元前方通常會出現虛擬環境名稱，例如：
```bash
(.venv) student@codespaces:~/project$
```
其中`(.venv)`表示目前正在使用名為 .venv 的虛擬環境。此時執行：`python --version`或`pip list`都將使用虛擬環境中的 Python 與套件，而不會影響系統中的其他 Python 專案。當工作完成後，可使用下列指令離開虛擬環境：`deactivate`。執行後，命令提示字元前方的`(.venv)`將會消失，表示已回到系統預設的 Python 環境。

本課程主要使用 GitHub Codespaces，因此通常只需要記住： 
```bash
python -m venv .venv 

source .venv/bin/activate 

deactivate
```

#### 課堂練習

請回答：

1 為什麼不建議將 .venv 目錄上傳到 GitHub？
因為 .venv 包含你電腦獨有的系統設定和龐大檔案，上傳後其他人換了電腦也無法使用，還會讓 GitHub 變得很慢。
2 為什麼建議將 requirements.txt 加入 Git 儲存庫？
因為 requirements.txt 是一張套件清單，將它上傳可以確保：

團隊環境一致：其他人下載專案後，能安裝完全相同的套件與版本，避免「在我電腦上可以執行，在你電腦上卻報錯」的問題。

方便部署：雲端伺服器（如 Docker、AWS、Heroku）能自動讀取它來安裝需要的環境。

檔案極小：它只是一行行純文字，幾 KB 的大小就能記錄整個專案的依賴關係。
3 requirements.txt 如何協助團隊合作？
消滅「在我電腦能跑」的爭議：明確記錄套件與精準版本號，確保所有成員安裝一模一樣的依賴，避免因版本差異導致程式碼出錯。

秒級重建環境：新成員加入專案時，只需執行 pip install -r requirements.txt 一行指令，幾秒鐘就能完成環境架設。

銜接自動化與部署（CI/CD）：雲端伺服器（如 GitHub Actions、Docker、AWS）能自動讀取這張清單安裝環境，實現自動化測試與線上部署。
4 如果兩位組員安裝了不同版本的 Flask，可能產生哪些問題？
語法與 API 不相容（程式碼直接崩潰）

Flask 跨大版本更新時會廢棄舊語法。例如舊版慣用的 before_first_request 裝飾器在 Flask 3.0 被徹底移除；新版寫法在舊版環境執行也可能跳出 AttributeError。

依存套件（Dependencies）版本衝突

Flask 依賴 Werkzeug、Jinja2、Click 等 底層套件。不同版本的 Flask 會拉下不同版本的 Werkzeug，可能導致底層運作邏輯（如 Request/Response 處理、Cookie 簽署或加密演算法）產生行為差異。

「在你電腦能跑，在我電腦不行」的除錯內耗

組員 A 推送（Push）程式碼到 Git 後，組員 B 拉取（Pull）下來卻無法啟動。雙方會花費大量時間檢查彼此的程式碼，最後才發現只是 Flask 版本不一致。

#### 課堂練習

- 為什麼團隊專案不一定適合隨時升級到最新版本？
程式碼壞掉：新版本常會刪除舊語法，導致原本正常的程式直接報錯。

其他套件卡住：專案用的其他套件可能還沒更新，造成版本彼此衝突。

多出未知 Bug：剛出的新版不一定穩定，舊版本反而經過時間驗證、更加安全。

- 為何需要將套件版本記錄在 requirements.txt 中？
避免自動裝到最新版而壞掉：如果不指定版本號，pip install 預設會下載最新版。若套件改版修改了語法，程式就會直接崩潰。

確保團隊與線上環境完全一致：讓每位開發者的電腦、測試伺服器與正式上線（Production）環境都使用一模一樣的套件組合，徹底消滅「在我電腦能跑」的問題。

實現一鍵快速架設：新加入的成員或自動化部署系統（CI/CD）只需執行 pip install -r requirements.txt，就能幾秒內還原出完整的開發環境。

- 更新套件後，為什麼要重新測試程式？
防止語法或 API 壞掉：套件升級時常會修改或廢棄舊功能（Breaking Changes），如果不測試，隱藏的錯誤可能在特定功能被觸發時才突然爆發。

避免底層行為改變：即使程式沒報錯，套件內部的演算法、預設參數或資料處理邏輯可能已有些微調整，導致運算結果與過去不符。

確保相容性與穩定度：套件更新可能同時引發其他相連套件的版本波動，透過測試才能確認整體系統依然穩定運作。

#### 移除套件

若要移除已安裝的套件，可使用：

```bash
pip uninstall package_name
```
移除套件時要注意，當移除某個套件時：`pip uninstall flask`，通常只會移除 Flask 本身，而不一定會移除所有相依套件。原因是他套件可能也正在使用這些相依套件。



#### 課堂練習

請回答：

為什麼安裝一個套件時，可能會同時安裝許多其他套件？
因為套件開發者會直接套用別人寫好的工具，這稱為「相依套件」（Dependencies）。

就像買一台冷氣，會附帶送來遙控器、安裝架和排氣管一樣——你雖然只點了 pip install 主要套件，pip 會自動把維持它正常運作所需的底層工具一併下載下來。

為什麼移除套件時需要注意相依性（Dependencies）？
怕誤刪（程式壞掉）：如果手動刪掉某個底層套件，其他沒刪的套件可能正需要它，導致程式跳出「找不到模組」而崩潰。

怕沒刪乾淨（留下垃圾）：用 pip uninstall 刪主套件時，當初順便裝進來的底層套件不會被自動刪除，長期下來會佔空間，甚至和未來的套件產生版本衝突。
為什麼使用虛擬環境後，可以更放心地安裝與移除套件？
不會搞垮系統或其他專案

所有套件都只存在該專案的 .venv 資料夾內。即使你不小心裝錯套件、把版本弄亂，影響範圍也僅限於該專案，電腦裡的其他專案與系統 Python 都完全不受影響。

具備「一鍵還原」的終極保險

萬一套件相依性亂掉或刪不乾淨，完全不需要花時間除錯。只要直接把 .venv 資料夾刪除（rm -rf .venv），重新建立一個全新的虛擬環境，幾秒鐘就能「砍掉重練」。

實驗成本極低，敢於嘗試

想測試新套件或大幅升級版本時，可以在虛擬環境裡隨意折騰。測試成功就把新套件寫入 requirements.txt；測試失敗就直接還原環境，完全沒有後顧之憂。

如果虛擬環境損壞或套件版本混亂，你會如何恢復開發環境？
刪除：直接把壞掉的 .venv 資料夾刪掉。

新建：輸入 python -m venv .venv 重新建立一個乾淨的環境。

還原：啟用環境後，執行 pip install -r requirements.txt 一鍵裝回所有套件。

[!WARNING]

不要隨意使用 sudo pip

在 Linux 或 macOS 系統中，除非非常清楚自己在做什麼，否則不建議使用：

sudo pip install package_name 這個指令可能將套件安裝到系統 Python 環境中， 並覆蓋或干擾作業系統原本的套件設定。

輕則造成 Python 環境混亂， 重則可能影響系統工具正常運作。

對於大多數 Python 專案而言， 使用虛擬環境（venv）即可解決套件安裝與版本管理問題。

建議流程如下：

- python -m venv .venv
- source .venv/bin/activate
- pip install package_name

如此套件只會安裝在目前專案的虛擬環境中， 不會影響系統中的其他 Python 專案。

在本課程使用的 GitHub Codespaces 環境中， 原則上不需要使用 sudo pip。

#### 課堂練習

請回答：

1. 為什麼不建議使用 `sudo pip install`？
搞垮作業系統：系統本身（如 Linux/macOS）有許多工具是用 Python 寫的。用 sudo 會覆蓋系統原有的套件，導致電腦功能或終端機直接癱瘓。

與系統更新衝突：系統的包管理員（如 apt）和 pip 會互相覆蓋彼此的檔案，導致未來更新電腦系統時直接卡死報錯。

資安風險極高：如果下載到惡意套件，sudo 權限會讓它直接取得你整台電腦的控制權，隨意刪除或偷竊檔案。
2. 虛擬環境（venv）如何降低套件安裝所帶來的風險？
隔離作業系統（保護電腦）

所有套件都只下載到專案資料夾內的 .venv 中，不會動到系統全域環境。即使裝到有問題的套件，也不會搞垮作業系統或觸發權限危機。

隔離其他專案（避免版本衝突）

每個專案都有各自獨立的 .venv。專案 A 用 Flask 2.0，專案 B 用 Flask 3.0，兩者互不干擾，徹底解決套件版本互相踩腳的問題。

提供「零成本」試錯機制（備份與復原）

測試新套件或升級失敗時，完全不需要除錯。只需直接刪除 .venv 資料夾並重新建立，幾秒鐘就能讓環境回到最乾淨的狀態。
3. 如果專案需要額外套件，應該安裝到哪裡？
保護電腦系統：套件只會裝在專案資料夾內（.venv），不會破壞作業系統本身的 Python 工具。

避免版本衝突：不同專案可以擁有各自獨立的虛擬環境與套件版本，互不干擾。

不怕嘗試與錯誤：如果套件裝壞或版本搞亂，只要直接刪除 .venv 資料夾重新建立，幾秒鐘就能一鍵還原。
4. 本課程使用 GitHub Codespaces，是否需要使用 `sudo pip`？為什麼？
不用權限就能裝：Codespaces 預設的使用者權限已經夠大，直接 pip install 就能順利安裝套件。

避免壞掉：用 sudo 可能會覆蓋掉容器內的底層工具，導致 VS Code 的自動補全或偵錯功能失靈。

養成好習慣：雲端環境只是幫你打包好開發工具，最標準做法依然是建立專案的 .venv 虛擬環境來管理套件！
