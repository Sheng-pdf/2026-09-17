## 第二部分：數學概念與 Python 型別

數學上有下列包含關係：

$$
\mathbb N

\subseteq

請回答：

1. Python 是否為上述每一個數系都提供獨立型別？

否。

Python 只為部分數學數系提供獨立型別：
整數 → int
實數（近似）→ float
複數 → complex
但：
自然數沒有獨立型別，使用 int
有理數沒有內建型別，通常使用 float 近似表示，或用 fractions.Fraction

2. 為什麼自然數與整數都使用 `int` 表示？

自然數本來就是整數的子集合。

3. Python 為什麼提供 `complex` 型別？

因為複數在科學計算、工程與數學中相當常見。

4. Python 為什麼沒有內建的有理數型別？

1.大部分程式只需要整數或浮點數。
2.有理數需要同時保存「分子」與「分母」，實作較複雜。
3.為了讓語言保持簡潔，因此沒有列入核心內建型別。

5. 下列三個物件的型別有何不同？

1 → 整數（int）
1.0 → 浮點數（float）
1+0j → 複數（complex）
自然數 ⊆ 整數 ⊆ 有理數 ⊆ 實數 ⊆ 複數

```python
1
1.0
1+0j
```

## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 任意精度整數，可超過 64 位元限制 |
| 整數 | `int` | PyLongObject | 使用可變長度儲存，不易溢位 |
| 實數 | `float` | PyFloatObject | 採用 IEEE 754 雙精度浮點數 |
| 複數 | `complex` | PyComplexObject | 由兩個 double（實部、虛部）組成 |
| 集合 | `set` | PySetObject | 基於雜湊表（Hash Table），元素不可重複 |
| 命題真假 | `bool` | PyBoolObject | 僅有 True 和 False，為 int 的子類別 |
| 串列 | `list` | PyListObject | 動態陣列（Dynamic Array），可存放任意型別物件 |

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？

因為 Python 的 int 採用「任意精度整數（Arbitrary Precision Integer）」設計，不受 CPU 的 32 位元或 64 位元整數大小限制。

2. Python 的 `int` 是否具有固定的位元數限制？

否。

3. 什麼是 Arbitrary Precision（任意精度）？

指數字需要多少位數，Python 就配置多少儲存空間。

4. CPython 使用什麼物件來實作 `int`？

PyLongObject
它將大整數拆成多個 digit（數字區塊）儲存。

5. Python 整數大小真正的限制是什麼？

理論上沒有上限。
實際限制來自：
電腦記憶體（RAM）
作業系統
Python 允許配置的最大物件大小
---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？

因為電腦使用二進位表示浮點數。
有些十進位小數無法被二進位精確表示。

2. 浮點數是否能精確表示所有小數？

不能。
3. 什麼是浮點數的精度（precision）？

能夠保留多少位有效數字。
Python 的 float 使用雙精度浮點數（64-bit）。

4. 什麼是浮點數的範圍（range）？

能表示的最小與最大數值。

5. 精度與範圍有何不同？

經度：數字有多精確
範圍：數字可以有多大或多小

具有 Unlimited Precision（無限制精度）。

1. Unlimited Precision 是什麼意思？

Python 的 int 不像 C、Java 等語言的整數具有固定的位元數（例如 32-bit 或 64-bit），而是可以根據需要自動增加儲存空間來表示更大的整數。

2. 它是否代表可以儲存無限大的整數？

不完全是。

3. CPython 如何實現這項特性？

PyLongObject
來實作 int。

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？

而大型 AI 模型卻通常選擇使用：
